# Fine-tuning Demo

This notebook outlines the intended fine-tuning flow for the MuseLoom model.

## Planned Flow

- Load metadata and training samples.
- Prepare prompts and conditioning labels.
- Build a `dreambooth_musicgen.py` command from MuseLoom config.
- Launch training and log checkpoints.
- Review generated outputs.

```python
import json
from pathlib import Path

config_path = Path('../configs/museloom_lora.json')
config = json.loads(config_path.read_text(encoding='utf-8'))
config
```

```python
command = ['python', 'dreambooth_musicgen.py']
for key, value in config.items():
    if isinstance(value, bool):
        if value:
            command.append(f'--{key}')
    elif value is not None:
        command.extend([f'--{key}', str(value)])

' '.join(command[:16]) + ' ...'
```
