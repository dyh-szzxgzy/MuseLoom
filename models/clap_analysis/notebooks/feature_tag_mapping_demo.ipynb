# Feature To Tag Mapping Demo

This notebook illustrates how low-level audio features can be translated into classroom-friendly tags.

```python
feature_sets = [
    {'tempo_bpm': 92, 'rms_energy': 0.214, 'label': 'calm_case'},
    {'tempo_bpm': 126, 'rms_energy': 0.412, 'label': 'bright_case'}
]
feature_sets
```

```python
mapped_tags = {
    'calm_case': ['平静', '温暖', '沉思'],
    'bright_case': ['激昂', '明亮', '推进感']
}
mapped_tags
```
